# 第7章：多模态LLM - 交互式学习

本notebook提供多模态LLM的交互式学习体验。

```python
from pathlib import Path
import os
import sys
import subprocess
import numpy as np
import matplotlib.pyplot as plt

if not Path("notebooks/bootstrap.py").exists():
    root = Path("/content/signal-to-intelligence")
    if not root.exists():
        subprocess.run(
            ["git", "clone", "https://github.com/datawhalechina/signal-to-intelligence.git", str(root)],
            check=True,
        )
    os.chdir(root)

if str(Path.cwd()) not in sys.path:
    sys.path.insert(0, str(Path.cwd()))

from notebooks.bootstrap import load_code_module

clip_similarity = load_code_module("code/ch07_multimodal_llm/clip_similarity.py")

%matplotlib inline

result = clip_similarity.run_experiment(d_model=128)
related_similarity = result['related_similarity']
unrelated_similarity = result['unrelated_similarity']

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

im = axes[0].imshow(related_similarity, cmap='YlOrRd', aspect='auto')
axes[0].set_xlabel('Text Index')
axes[0].set_ylabel('Image Index')
axes[0].set_title('Related Text-Image Similarity')
plt.colorbar(im, ax=axes[0])

means = [related_similarity.mean(), unrelated_similarity.mean()]
axes[1].bar(['Related', 'Unrelated'], means, color=['green', 'red'], alpha=0.7)
axes[1].set_ylabel('Mean Cosine Similarity')
axes[1].set_title('Similarity Comparison')
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("多模态LLM的发展趋势:")
print("- 图像和文本进入同一语义空间")
print("- 分辨率不断提高")
print("- 支持多图像理解")
print("- 中文优化")
print(f"CLIP检索准确率: {result['accuracy']:.2%}")
```
